```python
import pandas as pd
```

```python
df = pd.read_csv("Telco_customer_churn.csv")
```

```python
df.columns = df.columns.str.strip().str.replace("","_")
```

```python
cols_to_drop = [col for col in df.columns if "Churn" in col and col != "Churn_Value"]
df = df.drop(cols_to_drop, axis=1, errors='ignore')
```

```python
location_cols = ["City", "Zip_Code", "Latitude", "Longitude"]
df = df.drop([col for col in location_cols if col in df.columns], axis=1)
```

```python
df.to_csv("churn_dataset_tableau.csv", index=False)
```
